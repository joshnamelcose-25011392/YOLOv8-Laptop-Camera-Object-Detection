```python
import cv2
import matplotlib.pyplot as plt
from ultralytics import YOLO
import time

print("Libraries imported successfully!")
```

```python
model = YOLO("yolov8n.pt")

print("YOLOv8 model loaded successfully!")
```

```python
cap = cv2.VideoCapture(0)

if cap.isOpened():
    print("Laptop camera opened successfully!")
else:
    print("Error: Could not open camera.")
```

```python
print("Get ready...")
print("Capturing image in 5 seconds...")

time.sleep(5)

ret, frame = cap.read()

if ret:
    print("Image captured successfully!")
else:
    print("Failed to capture image.")
```

```python
# Convert BGR image to RGB
captured_image = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

# Display the captured image
plt.figure(figsize=(10, 6))
plt.imshow(captured_image)
plt.axis("off")
plt.title("Captured Image")
plt.show()
```

```python
# Perform YOLOv8 object detection
results = model(captured_image)

print("Object detection completed!")
```

```python
# Get the detected image with bounding boxes
detected_image = results[0].plot()

# Display the detected image
plt.figure(figsize=(10, 6))
plt.imshow(detected_image)
plt.axis("off")
plt.title("YOLOv8 Object Detection")
plt.show()
```

```python
cv2.imwrite("scene1_original.jpg", frame)

cv2.imwrite(
    "scene1_yolov8.jpg",
    cv2.cvtColor(detected_image, cv2.COLOR_RGB2BGR)
)

print("Scene 1 images saved successfully!")
```
